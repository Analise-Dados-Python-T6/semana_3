```python
logs_do_sistema = [
 "ERRO: Banco de dados desconectado as 03:00",
 {"produto": "Placa de Video RTX", "quantidade": 1, "preco": 3500.00, "status": "ok"},
 " mouse gamer , 5 , 45.90 ",
 "TESTE: admin, 1, 0.00",
 {"status": "erro", "mensagem": "Timeout no gateway de pagamento"},
 "Monitor Ultrawide; 1; 1500.00",
 {"produto": "Pendrive 32GB", "quantidade": 3, "preco": 25.00, "status": "ok"},
 "Cadeira de Escritorio, 2"
]

lista_final = []
lista_subtotal = []

#limpeza e separação dos logs
for log in logs_do_sistema:
  if type(log) == str:
    aux = log.replace(':', ';').replace(',', ';').split(';')

    if aux[0].lower() != 'erro' and aux[0].lower() != 'teste':
      if len(aux) == 3:
        lista_final.append(aux)

  elif type(log) == dict:
    if log.get('status') == 'ok':
      lista_final.append([log.get('produto'), log.get('quantidade'), log.get('preco')])

#montagem dos valores calculados de subtotal
try:
  for log in lista_final:
    subtotal = float(log[1]) * float(log[2])

    if subtotal > 1000:
      subtotal *= 0.9
      lista_subtotal.append(subtotal)

    elif subtotal > 100 and subtotal < 1000:
      lista_subtotal.append(subtotal)

    elif subtotal < 100:
      lista_subtotal.append(subtotal + 15)

    else:
      continue

except:
  print('erro')

print(lista_final)
print(lista_subtotal)
```
