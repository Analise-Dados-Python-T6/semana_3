```python
#desafio de loop em python

attempts = 0

correct_passwd = 'admin123'

while(attempts < 3):
  request_passwd = input('Senha: ')

  if request_passwd == correct_passwd:
    print('Senha correta, parabéns!')
    break

  else:
    attempts += 1
    print(f'Senha incorreta, vc tem {3 - attempts} tentativas')

print('Vecê ultrapassou o limite de tentativas, tente mais tarde')
```
