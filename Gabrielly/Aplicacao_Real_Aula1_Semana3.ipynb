# Aplicação real
Missão 1: Sistema de Login com Bloqueio de Segurança

Contexto da Vida Real:
Nenhum sistema seguro permite que um usuário tente adivinhar a senha infinitamente. Na nossa empresa, o padrão de segurança exige que a conta seja bloqueada após 3 tentativas erradas.

Seu Desafio:
Crie um script em Python usando o laço while que simule essa trava de segurança.

Regras do Sistema:

O usuário terá um limite máximo de 3 tentativas.
Peça para o usuário digitar a senha.
Se ele acertar a senha: Exiba a mensagem "Login efetuado com sucesso! Bem-vindo." e encerre o laço imediatamente.

Se ele errar a senha: Avise que a senha está incorreta e mostre quantas tentativas ele ainda tem.

Se as tentativas acabarem: Exiba um alerta "ALERTA: Conta bloqueada por excesso de tentativas." e o programa deve parar.


```python
senha_correta = "Sctec1008"
tentativas = 3

while tentativas > 0:
  senha= input("Digite sua senha: ")
  if senha != senha_correta:
        tentativas -= 1
        if tentativas == 0:
          print("ALERTA: Conta bloqueada por excesso de tentativas.")
        else:
          print(f"Senha incorreta. Você ainda tem {tentativas} tentativa(s) sua conta será bloqueada.")
  else:
        print("Login efetuado com sucesso! Bem-vindo.")
        break
```

# Aplicação real
Missão 2: Caixa Registradora (Acumulador de Dados)

Contexto da Vida Real:
Em análise de dados, muitas vezes recebemos uma enxurrada de informações e não sabemos exatamente quantas linhas de dados virão. Você foi encarregado de criar um script para o caixa de uma loja registrar as vendas do dia. O programa deve receber valores continuamente até que o caixa digite 0 para fechar o expediente.

Seu Desafio:
Crie um sistema que continue pedindo o valor da venda e somando tudo em um "acumulador". Quando o usuário digitar 0, o loop deve parar e exibir um relatório final.

Regras do Sistema:

Use o laço while para rodar enquanto o valor_venda for diferente de 0.
Dentro do loop, peça para o usuário digitar o valor da próxima venda.
Se o valor for diferente de 0: Some esse valor ao total arrecadado e adicione 1 na quantidade de vendas.
Se o valor for 0: O loop se encerrará naturalmente.
Imprima um relatório mostrando quantas vendas foram feitas no total e qual foi o faturamento total do dia.


```python
valor_venda = float(input("Digite o valor da venda: "))

faturamento = 0
quantidade_vendas = 0

while valor_venda != 0:

    faturamento = faturamento + valor_venda
    quantidade_vendas = quantidade_vendas + 1
    valor_venda = float(input("Digite o valor da próxima venda: "))


print("Relatório do dia")
print("Quantidade de vendas:", quantidade_vendas)
print("Faturamento total: R$", faturamento)
```
