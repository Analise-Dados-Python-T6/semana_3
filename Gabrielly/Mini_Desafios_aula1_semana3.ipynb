# Desafio 1
 Receba um nome com espaços e retorne-o em maiúsculas e sem espaços nas bordas.

```python
nome = input("Digite seu nome: ")

nome = nome.strip().upper()

print(nome)
```

# Desafio 2
 Receba a string "João,30,Recife" e exiba cada campo separado em uma linha diferente.

```python
dados = "joao,30,recife"

dados = dados.split(",")

for campo in dados:
    print(campo)
```

# Desafio 3
Conte quantas vogais existem em uma frase digitada pelo usuário, ignorando maiúsculas e minúsculas.

```python
frase = input("Digite uma frase: ")

quantidade_total_vogais = sum(frase.lower().count(vogal) for vogal in "aeiou")

print(quantidade_total_vogais)
```
