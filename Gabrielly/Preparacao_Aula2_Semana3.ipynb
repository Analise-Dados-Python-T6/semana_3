02

A partir da lista [3, 7, 1, 9, 4, 6, 2, 8, 5], crie uma lista apenas com os valores maiores que 5.

03

Dada a lista de nomes ['ana', 'pedro', 'maria', 'jose'], crie uma nova lista com os nomes em letras maiúsculas.

04

Crie uma lista com os múltiplos de 3 entre 1 e 30.

05

Dada lista de preços [100, 250, 80, 320, 150], aplique 10% de desconto em todos.


```python
numeros = list(range(1, 21))

print(numeros)
```

```python
numeros = [3, 7, 1, 9, 4, 6, 2, 8, 5]

maiores_que_5 = [numero for numero in numeros if numero > 5]

print(maiores_que_5)
```

```python
nomes = ['ana', 'pedro', 'maria', 'jose']

nomes_maiusculos = [nome.upper() for nome in nomes]

print(nomes_maiusculos)
```

```python
multiplos_3 = [numero for numero in range(1, 31) if numero % 3 == 0]

print(multiplos_3)
```

```python
precos = [100, 250, 80, 320, 150]

precos_com_desconto = [preco * 0.9 for preco in precos]

print(precos_com_desconto)
```

# Parte 2

06

Dada a string 'Python é incrível', crie uma lista com cada palavra que tem mais de 4 letras.

07

Crie uma lista que substitui números negativos por 0, mantendo os positivos. Lista: [-3, 5, -1, 8, -7, 2].

08

Dada a lista de strings ['123', 'abc', '456', 'def', '789'], crie uma lista apenas com os valores que são números (use .isdigit()).

09

A partir de uma lista de temperaturas em Celsius [0, 15, 22, 30, 38], converta para Fahrenheit apenas as temperaturas acima de 20°C.

10

Crie uma lista de tuplas (número, quadrado, cubo) para os números de 1 a 5.


```python
texto = 'Python é incrível'

palavras = [palavra for palavra in texto.split() if len(palavra) > 4]

print(palavras)
```

```python
numeros = [-3, 5, -1, 8, -7, 2]

resultado = [numero if numero >= 0 else 0 for numero in numeros]

print(resultado)
```

```python
valores = ['123', 'abc', '456', 'def', '789']

numeros = [valor for valor in valores if valor.isdigit()]

print(numeros)
```

```python
temperaturas = [0, 15, 22, 30, 38]

fahrenheit = [temperatura * 9/5 + 32 for temperatura in temperaturas if temperatura > 20]

print(fahrenheit)
```

```python
resultado = [(numero, numero**2, numero**3) for numero in range(1, 6)]

print(resultado)
```
