## As Regras de Negócio:

A Função de Limpeza (limpar_transacao). Crie uma função que receba um registro (podendo ser string ou dicionário). Ela deve:
Extrair e padronizar o tipo (maiúsculo), o valor (float), o status (minúsculo) e a hora (apenas o número inteiro, ex: de "14:30" extrair 14).
Se a string usar ; como separador, corrija para ,.
Regra de Anomalia: Se o valor da transação for negativo, a função deve retornar None.
Caso contrário, deve retornar um dicionário padronizado: {"tipo": tipo, "valor": valor, "status": status, "hora": hora}.

O Filtro Rápido (List Comprehension).
Usando apenas uma linha de código, crie uma lista chamada transacoes_validas. Ela deve passar todos os dados brutos pela função de limpeza e guardar apenas os que não retornaram None e cujo status seja diferente de "erro".

3. A Função de Risco (avaliar_risco). Crie uma função que receba o valor, o tipo e a hora. Aplique as regras:
Horário Noturno (22h às 05h):

PIX acima de R$ 1.000 ➔ ALTO

Horário Diurno (06h às 21h):
Qualquer tipo acima de R$ 50.000 ➔ CRÍTICO
PIX acima de R$ 5.000 ➔ ALTO
TED acima de R$ 10.000 ➔ ALTO
Caso contrário: ➔ BAIXO

Relatório Final - Faça um loop pela sua lista de transacoes_validas e aplique a função de risco. Exiba na tela apenas as transações que representam risco ALTO ou CRÍTICO.

```python
batch_de_transacoes = [
    "PIX, 1500.50, aprovado, 14:30",
    "TED ; 200.00 ; erro ; 10:15",
    {"tipo": "DOC", "valor": 8500.00, "status": "aprovado", "horario": "11:00"},
    "PIX, -50.00, aprovado, 08:00",
    {"tipo": "PIX", "valor": 1500.00, "status": "aprovado", "horario": "02:30"},
    "  TED,   55000.00  , aprovado  , 15:00",
    {"tipo": "BOLETO", "valor": 150.00, "status": "erro", "horario": "19:00"},
    "PIX ; 25000.00 ; aprovado ; 23:15"
]

#Limpar dados

def limpar_transacao(registro):

  if isinstance(registro, str):
    registro = registro.replace(";", ",") #Padronizar separador para ,
    dados = registro.split(",")

    tipo = dados[0].strip().upper() #remove espaços e coloca em maisculo os dados do primeiro elemento
    valor = float(dados[1].strip()) #remove espaços e transforma em numero decimal do segundo elemento
    status = dados[2].strip().lower() #remove espaços e coloca em minusculo terceiro elemento
    hora = int(dados[3].strip().split(":")[0]) #remove espaços separa por : e pega somente o primeiro elemento

  elif isinstance(registro,dict):
    tipo = registro["tipo"].upper()
    valor = float(registro["valor"])
    status = registro["status"].lower()
    hora = int(registro["horario"].split(":")[0])

  if valor < 0:
    return None

  return {
        "tipo": tipo,
        "valor": valor,
        "status": status,
        "hora": hora
    }

  #Filtro rapido com List Comprehension

transacoes_validas = [limpar_transacao(transacao) for transacao in batch_de_transacoes if limpar_transacao(transacao) is not None and limpar_transacao(transacao)["status"] != "erro"]

#Avaliar risco

def avaliar_risco(valor, tipo, hora):

    if hora >= 22 or hora <= 5:

        if valor > 20000:
            return "CRÍTICO"

        elif tipo == "PIX" and valor > 1000:
            return "ALTO"

    else:

        if valor > 50000:
            return "CRÍTICO"

        elif tipo == "PIX" and valor > 5000:
            return "ALTO"

        elif tipo == "TED" and valor > 10000:
            return "ALTO"

    return "BAIXO"

    # Relatório Final

for transacao in transacoes_validas:

    risco = avaliar_risco(
        transacao["valor"],
        transacao["tipo"],
        transacao["hora"]
    )

    if risco == "ALTO" or risco == "CRÍTICO":
        print(f"Tipo: {transacao['tipo']}")
        print(f"Valor: R$ {transacao['valor']:.2f}")
        print(f"Status: {transacao['status']}")
        print(f"Hora: {transacao['hora']}h")
        print(f"Risco: {risco}")
        print("-" * 30)
```
