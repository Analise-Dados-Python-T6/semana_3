```python
def saudacao(nome):
  return f"Olá, {nome}!"

print(saudacao("Gabrielly"))

```

```python
def dobro(num):
  return num*2

print(dobro(5))
```

```python
def numero_par(numero):
 return numero % 2 == 0

print(numero_par(3))
```

```python
def limpar_texto(palavra):
  return palavra.strip().lower()

print(limpar_texto("  TEST "))
```

```python
def calcular_media(numeros):
  return sum(numeros) / len(numeros)

notas = [8,7,5,6]

print(calcular_media(notas))
```
